```python
from src.utils import data_utils
import pandas as pd
```

```python
data = data_utils.load_data("../data/interim/eda_1_0_1_df.pkl")
```

```python
fe_data = data.copy()
```

```python
"""
Cell generated by Data Wrangler.
"""
def clean_data(fe_data):
    # Rename column ('participants_age', 'modified_dem_0110') to 'age'
    fe_data = fe_data.rename(columns={('participants_age', 'modified_dem_0110'): 'age'})
    # Rename column ('participants_sex', 'dem_0500') to 'sex'
    fe_data = fe_data.rename(columns={('participants_sex', 'dem_0500'): 'sex'})
    # Rename column ('height_in_inches', 'dem_0610') to 'height_inches'
    fe_data = fe_data.rename(columns={('height_in_inches', 'dem_0610'): 'height_inches'})
    # Rename column ('weight_in_pounds', 'dem_0700') to 'weight_pounds'
    fe_data = fe_data.rename(columns={('weight_in_pounds', 'dem_0700'): 'weight_pounds'})
    # Rename column ('body_mass_index_(bmi)', 'dem_0800') to 'bmi'
    fe_data = fe_data.rename(columns={('body_mass_index_(bmi)', 'dem_0800'): 'bmi'})
    return fe_data

fe_data_clean = clean_data(fe_data.copy())
fe_data_clean.head()
```

```python
"""
Cell generated by Data Wrangler.
"""
def clean_data(fe_data_clean):
    # Rename column ('number_of_people_living_in_your_household', 'bthbts_0500') to 'num_people_in_house'
    fe_data_clean = fe_data_clean.rename(columns={('number_of_people_living_in_your_household', 'bthbts_0500'): 'num_people_in_house'})
    # Drop columns: '('number_of_children_aged_5_years_or_younger_living_in_your_household', 'bthbts_0510')', '('number_of_children_aged_6_17_living_in_your_household', 'bthbts_0520')' and 2 other columns
    fe_data_clean = fe_data_clean.drop(columns=[('number_of_children_aged_5_years_or_younger_living_in_your_household', 'bthbts_0510'), ('number_of_children_aged_6_17_living_in_your_household', 'bthbts_0520'), ('number_of_adults_aged_18_59_living_in_your_household', 'bthbts_0530'), ('number_of_adults_aged_60_years_of_older_living_in_your_household', 'bthbts_0540')])
    return fe_data_clean

fe_data_clean_1 = clean_data(fe_data_clean.copy())
fe_data_clean_1.head()
```

```python
""" 
encoding work/school start/end time
turn it into number of hours before/after 9am/5pm 
"""

work_start_times = fe_data_clean_1[('time_school_or_work_starts', 'sched_9910')].dt.time
work_end_times = fe_data_clean_1[('time_school_or_work_ends', 'sched_9920')].dt.time


nine_am = pd.to_datetime("09:00:00").time()
five_pm = pd.to_datetime("17:00:00").time()

def time_diff_in_hours(time1, time2):
    datetime1 = pd.to_datetime(time1, format='%H:%M:%S')
    datetime2 = pd.to_datetime(time2, format='%H:%M:%S')
    diff = (datetime1 - datetime2).total_seconds() / 3600
    return diff

fe_data_clean_1[('work_start_time_diff_from_9am', 'sched_9910')] = work_start_times.apply(lambda x: time_diff_in_hours(x, nine_am))
fe_data_clean_1[('work_end_time_diff_from_5pm', 'sched_9920')] = work_end_times.apply(lambda x: time_diff_in_hours(x, five_pm))

```

```python
"""
Cell generated by Data Wrangler.
"""
def clean_data(fe_data_clean_1):
    # Rename column ('time_school_or_work_starts', 'sched_9910') to 'sw_9am_diff'
    fe_data_clean_1 = fe_data_clean_1.rename(columns={('time_school_or_work_starts', 'sched_9910'): 'sw_9am_diff'})
    # Rename column ('time_school_or_work_ends', 'sched_9920') to 'sw_5pm_end_diff'
    fe_data_clean_1 = fe_data_clean_1.rename(columns={('time_school_or_work_ends', 'sched_9920'): 'sw_5pm_end_diff'})
    # Rename column 'sw_9am_diff' to 'sw_9am_start_diff'
    fe_data_clean_1 = fe_data_clean_1.rename(columns={'sw_9am_diff': 'sw_9am_start_diff'})
    return fe_data_clean_1

fe_data_clean_2 = clean_data(fe_data_clean_1.copy())
fe_data_clean_2.head()
```

```python
fe_data_clean_2['sw_9am_start_diff'] = fe_data_clean_2[('work_start_time_diff_from_9am', 'sched_9910')]
fe_data_clean_2['sw_5pm_end_diff'] = fe_data_clean_2[('work_end_time_diff_from_5pm', 'sched_9920')]

fe_data_clean_2 = fe_data_clean_2.drop(columns=[('work_start_time_diff_from_9am', 'sched_9910'), ('work_end_time_diff_from_5pm', 'sched_9920')])
```

```python
def encode_gender(df):
    transformed_df = df.copy()
    transformed_df["sex"] = transformed_df["sex"].map({"M": 0, "F": 1})
    return transformed_df
```

```python
def encode_excercise_amt_or_time(df):
    transformed_df = df.copy()
    transformed_df["exercise_amt_or_time"] = transformed_df["exercise_amt_or_time"].map({
        "rarely_or_never" : 0,
        "between dinner and bedtime" : 1,
        "late morning" : 1, 
        "early morning" : 1,
        "early afternoon" : 1,
        "late afternoon" : 1,
    })
    return transformed_df
```

```python
import numpy as np

# should make 0 no naps 1 weekly/monthyl naps
def nap_encoder(df):
    transformed_df = df.copy()
    transformed_df["nap_num"] = np.where(
        transformed_df["nap_num"].str.contains("per week"),
        1,
        np.where(
            transformed_df["nap_num"].str.contains("per month"),
            0,
            np.where(transformed_df["nap_num"].str.contains("per year"), 0, 0),
        ),
    )
    transformed_df = transformed_df.rename(columns={"nap_num": "frequent_naps"})
    return transformed_df
```

```python
def boolean_converter(df, column_name):
    transformed_df = df.copy()
    if isinstance(column_name, list):
        for col in column_name:
            if transformed_df[col].dtype == "boolean":
                transformed_df[col] = transformed_df[col].fillna(False).astype(int)
            else:
                print(f"Column {col} is not of boolean type.")
    elif transformed_df[column_name].dtype == "boolean":
        transformed_df[column_name] = transformed_df[column_name].fillna(False).astype(int)
    else: 
        print(f"Column {column_name} is not of boolean type.")
    return transformed_df
```

```python
def encode_daytime_sleepiness(df):
    transformed_df = df.copy()
    transformed_df["difficulty_staying_awake_during_the_day_number_of_times"] = (
        np.where(
            transformed_df[
                "difficulty_staying_awake_during_the_day_number_of_times"
            ].str.contains("Day"),
            1,
            np.where(
                transformed_df[
                    "difficulty_staying_awake_during_the_day_number_of_times"
                ].str.contains("Week"),
                1,
                0,
            ),
        )
    )
    transformed_df = transformed_df.rename(columns={"difficulty_staying_awake_during_the_day_number_of_times": "frequent_daytime_sleepiness"})
    return transformed_df
```

```python
def remember_dreams_encoder(df):
    transformed_df = df.copy()
    dream_count = transformed_df["remember_dreams_times"].str.extract(r'(\d+)').astype(float)
    dream_time_frame = transformed_df["remember_dreams_times"].str.extract(r'per\s(\w+)').iloc[:, 0]
    
    transformed_df["remember_dreams_times"] = np.where(
        (dream_count >= 3) & ((dream_time_frame == "week") | ((dream_time_frame == "month") & (dream_count >= 12))),
        "frequent",
        np.where(
            (dream_count < 3) & ((dream_time_frame == "week") | ((dream_time_frame == "month") & (dream_count < 12))),
            "infrequent",
            "rarely_or_never"
        )
    )
    transformed_df = transformed_df.rename(columns={"remember_dreams_times": "dream_recall_frequency"})

    return transformed_df
```

```python
# TODO: find way to handle -55 and other placeholder NAs 
def placeholder_remover(df, column_name, placeholder=-55):
    transformed_df = df.copy()
    transformed_df[column_name] = transformed_df[column_name].replace(placeholder, pd.NA)
    return transformed_df
```

```python
def ahi_binary(df, threshold=5):
    transformed_df = df.copy()
    transformed_df["ahi_binary"] = (transformed_df["ahi"] >= threshold).astype(int)
    return transformed_df
```

```python
# Parallel Analysis to determine optimal number of factors
# Compare observed eigenvalues to eigenvalues from random data

n_iterations = 100
n_obs, n_vars = features_df.shape

# Observed eigenvalues from correlation matrix
fa_all = FactorAnalyzer(rotation=None, n_factors=n_vars, method='principal')
fa_all.fit(features_df)
observed_ev, _ = fa_all.get_eigenvalues()

# Generate random eigenvalues
random_evs = np.zeros((n_iterations, n_vars))
for i in range(n_iterations):
    random_data = np.random.normal(size=(n_obs, n_vars))
    random_corr = np.corrcoef(random_data, rowvar=False)
    random_evs[i, :] = np.sort(np.linalg.eigvalsh(random_corr))[::-1]

mean_random_ev = random_evs.mean(axis=0)
p95_random_ev = np.percentile(random_evs, 95, axis=0)

# Find number of factors where observed > 95th percentile of random
n_factors_parallel = np.sum(observed_ev > p95_random_ev)

# Plot
fig, ax = plt.subplots(figsize=(10, 6))
k = min(30, n_vars)  # plot up to 30 components for readability
ax.plot(range(1, k + 1), observed_ev[:k], 'b-o', label='Observed eigenvalues')
ax.plot(range(1, k + 1), mean_random_ev[:k], 'r--^', label='Mean random eigenvalues')
ax.plot(range(1, k + 1), p95_random_ev[:k], 'g--s', label='95th percentile random')
ax.axhline(y=1, color='gray', linestyle=':', alpha=0.5, label='Kaiser criterion (EV=1)')
ax.axvline(x=n_factors_parallel, color='purple', linestyle='--', alpha=0.7,
           label=f'Suggested factors: {n_factors_parallel}')
ax.set_xlabel('Factor Number')
ax.set_ylabel('Eigenvalue')
ax.set_title('Parallel Analysis Scree Plot')
ax.legend()
plt.tight_layout()
plt.show()

print(f"\nParallel analysis suggests {n_factors_parallel} factors")
print(f"(retain factors where observed eigenvalue > 95th percentile of random)")
```

```python
def parallel_analysis(df, n_iterations=100):
    n_obs, n_vars = df.shape

    # Observed eigenvalues from correlation matrix
    fa_all = FactorAnalyzer(rotation=None, n_factors=n_vars, method='principal')
    fa_all.fit(df)
    observed_ev, _ = fa_all.get_eigenvalues()

    # Generate random eigenvalues
    random_evs = np.zeros((n_iterations, n_vars))
    for i in range(n_iterations):
        random_data = np.random.normal(size=(n_obs, n_vars))
        random_corr = np.corrcoef(random_data, rowvar=False)
        random_evs[i, :] = np.sort(np.linalg.eigvalsh(random_corr))[::-1]

    mean_random_ev = random_evs.mean(axis=0)
    p95_random_ev = np.percentile(random_evs, 95, axis=0)

    # Find number of factors where observed > 95th percentile of random
    n_factors_parallel = np.sum(observed_ev > p95_random_ev)

    # Plot
    fig, ax = plt.subplots(figsize=(10, 6))
    k = min(30, n_vars)  # plot up to 30 components for readability
    ax.plot(range(1, k + 1), observed_ev[:k], 'b-o', label='Observed eigenvalues')
    ax.plot(range(1, k + 1), mean_random_ev[:k], 'r--^', label='Mean random eigenvalues')
    ax.plot(range(1, k + 1), p95_random_ev[:k], 'g--s', label='95th percentile random')
    ax.axhline(y=1, color='gray', linestyle=':', alpha=0.5, label='Kaiser criterion (EV=1)')
    ax.axvline(x=n_factors_parallel, color='purple', linestyle='--', alpha=0.7,
               label=f'Suggested factors: {n_factors_parallel}')
    ax.set_xlabel('Factor Number')
    ax.set_ylabel('Eigenvalue')
    ax.set_title('Parallel Analysis Scree Plot')
    ax.legend()
    plt.tight_layout()
    plt.show()

    print(f"\nParallel analysis suggests {n_factors_parallel} factors")
```

```python
parallel_analysis(features_df_2)
```
